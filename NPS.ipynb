### EXPLORANDO OS FATORES QUE MAIS IMPACTAM NO ÍNDICE DE SATISFAÇÃO DO CLIENTE (NPS)


---


**● Qual problema de negócio está sendo resolvido**- •	Os dados de NPS são coletados após a finalização de todo o processo de compra, o que implica em falta de poder de tomada de decisão voltado para evitar a ocorrência dos problemas, já que somente se tem a visibilidade após o problema já ter acontecido.



**● Por que o NPS é importante para um e-commerce?**


---


O indicador de experiência do cliente, é uma métrica que valida a satisfação entre os clientes através de pesquisas. Onde o cliente que apresenta estar muito satisfeito com a marca, tende a manter fidelidade e promover a marca para que outras pessoas possam se tornar possíveis clientes. Ao mesmo tempo em que, clientes que se apresentam insatisfeitos, tendem a não utilizar mais da marca e até mesmo difamar para o público.
Tendo um NPS bem claro, é possível estudar abordagens buscando alternativas para contornar casos de insatisfação por parte dos clientes, visando uma reversão nos possíveis casos de perda de clientela por experiências ruins.


**● Quais áreas poderiam se beneficiar desses insights? Exemplos:**


---



•	**Logística**: A logística através desses insights, pode reanalisar todo seu processo em busca de entender como melhorar a satisfação do cliente final.
Essa reestruturação, pode se dar no processo de expedição, no transporte das mercadorias, no desenho das rotas, nos prazos de entrega e até mesmo na análise das transportadoras, com possibilidade de identificar um possível gargalo no processo.

•	**Produto/Fábrica:** Pode identificar falhas no processo de produção que podem ser convertidas em melhorias que permitam a criação de produtos com maior qualidade ou que agradem as exigências por parte do cliente.

•	**Pricing:** Com o indicador NPS e feedbacks dos clientes, a área de pricing pode rever seus preços de venda, e identificar se o valor cobrado, é justo, condiz com o mercado, ou está sendo considerado incorretamente.

## Buscando os valores mínima, média e máxima para cada coluna

```python
df_nps[["idade_cliente", "meses_como_cliente", "valor_pedido",
    "quantidade_itens", "valor_desconto", "parcelas_pagamento",
    "tempo_entrega_dias", "dias_atraso_entrega", "tentativas_entrega",
    "tempo_resolucao_dias", "nota_nps"]].agg(["min", "mean", "max"]).round(2).T
```

```python
df_nps["classificacao_nps"] = 'neutro'
df_nps.loc[df_nps['nota_nps'] <=6, "classificacao_nps"] = 'detrator'
df_nps.loc[df_nps['nota_nps'] >= 9, "classificacao_nps" ] = 'promotor'
```

```python
df_nps.sort_values("nota_nps", ascending=True).head(5)
```

```python
colunas_numericas = ["idade_cliente", "meses_como_cliente", "valor_pedido",
    "quantidade_itens", "valor_desconto", "parcelas_pagamento",
    "tempo_entrega_dias", "dias_atraso_entrega", "tentativas_entrega",
    "contatos_sac", "tempo_resolucao_dias"]
```

```python
nps_geral = df_nps.groupby("classificacao_nps")[colunas_numericas].mean()
display(nps_geral)
```
